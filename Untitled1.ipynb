```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
from pathlib import Path
import textwrap
import cga_utils
import math
from typing import Iterable, List, Optional, Tuple
import hdbscan
from sklearn.feature_extraction import DictVectorizer
import cluster_features

from sklearn.preprocessing import StandardScaler
errors = pd.read_csv('res/e38_23.csv').query('exact_match == False')
vects = errors[["calc_pattern", "code_calc_pattern", "scale", "pred_scale"]]
X_dict = vects.to_dict(orient="records")
dv = DictVectorizer(sparse=False)
X_features = dv.fit_transform(X_dict).astype(int)

clusterer = hdbscan.HDBSCAN(min_cluster_size=4, min_samples=2, metric="hamming", cluster_selection_method="eom")

clusterer.fit_predict(X_features)

labels = clusterer.labels_
probs = clusterer.probabilities_
persistence = getattr(clusterer, "cluster_persistence_", None)

print("Klaszterek száma (zaj nélkül):", len(set(labels) - {-1}))
print("Zaj arány:", np.mean(labels == -1))
print("Prob. min/mean/max:", probs.min(), probs.mean(), probs.max())
if persistence is not None:
    print("Persistence egyediek:", np.unique(persistence), " | len:", len(persistence))
```
