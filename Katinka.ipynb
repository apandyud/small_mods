```python
# Ábra készítése feliratokkal (magyar)
from adjustText import adjust_text
import matplotlib.pyplot as plt

# Első adatsor (kékkel felülről) — címkék + értékek
labels_blue = [
    ("nő", 0.6),
    ("férfi", 0.84),
    ("házas", 0.5),
    ("elvált", 1.4),
    ("nincs gyermeke", 0.71),
    ("2 gyermeke van", 0.62),
    ("alacsonyan iskolázott", 0.58),
    ("közepesen iskolázott", 0.79),
    ("község", 0.42),
    ("Budapest", 1.14),
    ("15–29 éves", 0.7),
    ("30–49 éves", 0.91),
    ("50+", 0.37),
]

# Második adatsor (pirossal alulról) — címkék + értékek
labels_red = [
    ("nem tag", 0.87),
    ("aktív tag", -0.74),
    ("katolikus", 0.43),
    ("protestáns", 0.61),
    ("legalább havi\ntemplombajáró", -0.57),
    ("nem jár templomba", 1.3),
    ("ÁTLAG", 0.71),
]

# Beállítások és adatok
all_vals = [v for _, v in labels_blue + labels_red]

plt.figure(figsize=(16, 6))
plt.axhline(0, color="black", linewidth=1)  # számegyenes

ax = plt.gca()   # 
ax.spines['top'].set_visible(False)
#ax.spines['left'].set_visible(False)
#ax.spines['right'].set_visible(False)
#ax.spines['bottom'].set_visible(False)
#ax.spines['bottom'].set_position(('data', 0))
ax.xaxis.set_label_position('bottom')
ax.tick_params(axis='x', labelsize=12) 

shifts = {
    "2 gyermeke van": 0.023,
    "alacsonyan iskolázott" : -0.025,
    "nincs gyermeke" : 0.025,
    "15–29 éves": -0.0,
    
}
unders = ["Budapest", "protestáns", "50+"]



# Kék nyilak felülről + feliratok
for label, val in labels_blue:
    plt.arrow(val, 0, 0, 0.06, head_width=0.02, head_length=0.02, fc='blue', ec='blue', length_includes_head=True)
    
    if label in shifts:
        shift = shifts[label]        
    else:
        shift = 0        
    
        
    txt = plt.text(
        val -.02 + shift, 0.07, f"{label}",
        ha='left', va='bottom', fontsize=12, rotation=45
    )
    #texts.append(txt)

# Piros nyilak alulról + feliratok
for label, val in labels_red:
    plt.arrow(val, 0, 0, -0.06, head_width=0.02, head_length=0.02, fc='red', ec='red', length_includes_head=True)
    
    if label in shifts:
        shift = shifts[label]        
    else:
        shift = 0 
        
    txt = plt.text(
        val + shift, -0.07, f"{label}",
        ha='right', va='top', fontsize=12,  rotation=45
    )
    #texts.append(txt)

#adjust_text(texts, only_move={'points':'y', 'text':'y'}, arrowprops=dict(arrowstyle="->", color='gray'))

# Tengelyek és feliratok
x_pad = 0.2
plt.xlim(min(all_vals) - x_pad, max(all_vals) + x_pad)
plt.ylim(-0.22, 0.22)
plt.xlabel("Tradicionális  ←   →  Szekuláris-racionális", fontsize=15)
plt.yticks([])

#plt.tight_layout()

# Mentés és megjelenítés
out_path = "szamegyenes_feliratokkal.png"
plt.savefig(out_path, dpi=200, bbox_inches="tight")
plt.show()

out_path
```
