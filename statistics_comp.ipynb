```python
from scipy import stats

# Feltételezzük, hogy mindkét DataFrame azonos sorrendben tartalmazza a kérdéseket
# és minden cella 1 (helyes) vagy 0 (helytelen).
# Példa: df_complex["correct"] és df_simple["correct"] az oszlopok
def calc_t_stats(res1,res2):
    complex_correct = res1["em"].astype(int)
    simple_correct = res2["em"].astype(int)
    
    # Paired t-test
    t_stat, p_value = stats.ttest_rel(simple_correct, complex_correct)
    
    print(f"t-statistic: {t_stat:.4f}")
    print(f"p-value: {p_value:.6f}")
    
    # Szignifikancia ellenőrzése 5%-os szinten
    if p_value < 0.05:
        print("Significant difference between complex and simple prompts.")
    else:
        print("No significant difference between complex and simple prompts.")
```

```python
for pair in pairs:
    print(pair["name"])
    calc_t_stats(pair["complex"], pair["simple"])
```

```python
res = []
for pair in pairs:
    diff = pair["complex"]["em"].astype(int) - pair["simple"]["em"].astype(int)
    
    # Összesítés
    summary = {
        "n_questions": len(diff),
        "n_positive_diff": (diff > 0).sum(),    # egyszerű prompt jobb
        "n_negative_diff": (diff < 0).sum(),    # komplex prompt jobb
        "n_equal": (diff == 0).sum(),           # nincs különbség
        "mean_diff": diff.mean(),
        "std_diff": diff.std(ddof=1),
        "prop_positive": (diff > 0).mean()      # arány, amikor az egyszerű prompt jobb
    }
    res.append(summary)
pd.DataFrame(res)
```
